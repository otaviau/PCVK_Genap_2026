##2. Tampilkan image dalam channel Red-Blue dan Green-Blue saja!

```python
#Load gambar & konversi BGR ke RGB/
img_orig = cv2.imread("/content/drive/MyDrive/PCVK/Gambar/contohktm.png")
img_rgb = cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB)

#Channel Red-Blue saja (Green diset ke 0)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

#Channel Green-Blue saja (Red diset ke 0)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title("Channel Red-Blue Saja")

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title("Channel Green-Blue Saja")

plt.show()
```

##5. Buat 5 kotak berbagai ukuran dan warna yang berbeda dalam satu image (disarankan menggunakan bilangan acak/random)!

```python
#membuat kanvas hitam ukuran 512x512
canvas_kotak = np.zeros((512, 512, 3), dtype=np.uint8)

# 5 kotak secara acak
for i in range(5):
    x1, y1 = np.random.randint(0, 400, size=2)
    x2, y2 = x1 + np.random.randint(30, 100), y1 + np.random.randint(30, 100)

    color = tuple(map(int, np.random.randint(0, 256, size=3)))

    thickness = np.random.choice([-1, 2, 4, 6]) #-1 berarti terisi penuh (fill)

    cv.rectangle(canvas_kotak, (x1, y1), (x2, y2), color, thickness)

plt.imshow(canvas_kotak)
plt.title("5 Kotak Warna & Ukuran Acak")
plt.show()
```

##7. Buat rectangle dan circle pada bagian wajah dari image foto anda saat beraktifitas (bukan pasfoto).

```python
foto_aktivitas = cv2.imread("/content/drive/MyDrive/PCVK/Gambar/contohktm.png")
foto_rgb = cv2.cvtColor(foto_aktivitas, cv2.COLOR_BGR2RGB)

#koordinat titik wajah
x, y, w, h = 155, 400, 120, 120

#rectangle di sekeliling wajah
cv.rectangle(foto_rgb, (x, y), (x + w, y + h), (0, 255, 0), 3)

#circle di pusat wajah
center_x = x + w // 2
center_y = y + h // 2
radius = w // 2
cv.circle(foto_rgb, (center_x, center_y), radius, (255, 0, 0), 2)

plt.imshow(foto_rgb)
plt.title("Deteksi Wajah Manual (Rectangle & Circle)")
plt.show()
```

##8. Buat rectangle pada bagian sudut bawah kiri channel B pada color space RGB dari citra (kitten/ lena/ mandrill/ male/ female/ couple/ sailboat/ peppers)!

```python
img_peppers = cv2.imread("/content/drive/MyDrive/PCVK/Gambar/contohktm.png")
img_peppers_rgb = cv2.cvtColor(img_peppers, cv2.COLOR_BGR2RGB)

blue_channel = img_peppers_rgb[:, :, 2].copy()

h_img, w_img = blue_channel.shape

pt1 = (10, h_img - 80)
pt2 = (100, h_img - 10)

cv.rectangle(blue_channel, pt1, pt2, (255), 3)

plt.imshow(blue_channel, cmap='gray')
plt.title("Rectangle di Sudut Bawah Kiri (Channel Blue)")
plt.show()
```
